#### $Diagnostico$ $y$ $propuesta$ $de$ $solución$


<div align="justify">
El proceso de gestión de peritajes en el banco se encuentra limitado por una carga operativa excesivamente manual y dependiente de la intervención humana en tareas de bajo valor. La recepción de documentos escaneados en formato PDF obliga al personal a realizar extracciones de datos de forma tradicional, lo cual ralentiza la migración de información hacia los datasets preliminares y aumenta exponencialmente el riesgo de errores de digitación o inconsistencias en los registros. Esta fragmentación operativa impide la consolidación ágil de un Data Frame Maestro, convirtiéndose en un cuello de botella crítico que afecta la integridad del repositorio central y demora significativamente la validación de los peritajes entregados por los proveedores externos. En última instancia, esta carencia de automatización se traduce en tiempos de respuesta ineficientes que impactan la productividad y la capacidad de reacción analítica de la entidad financiera frente a sus procesos de control.
</div>    

<div align="justify">
Como propuesta de solución y optimización para mitigar estas deficiencias, se propone una reingeniería basada en la implementación de un flujo ETL automatizado que integre tecnologías de OCR y librerías avanzadas de Python para la captura inteligente de datos. Esta solución permitirá la digitalización e ingesta inmediata de los documentos PDF, estandarizando la extracción, limpieza y carga de información directamente en el Data Frame Maestro sin necesidad de procesos manuales intermedios. Al centralizar esta información depurada, se facilitará el despliegue de tableros de control dinámicos mediante Streamlit, permitiendo una visualización analítica e interactiva de los resultados y métricas de peritaje en tiempo real. Este enfoque no solo optimizará drásticamente los tiempos de procesamiento y entrega, sino que también dotará al banco de una estructura de datos escalable que garantiza la trazabilidad de los proveedores y mejora la toma de decisiones basada en evidencia.
</div>


---------------------------
 #### $Objetivos$

<div align="justify">
- Construir una herramienta en Python para reducir la fricción y operatividad manual entre proveedores del servicio de peritaje y optimizando los tiempos de respuesta a los usuarios deudores, como también a la construcción de informes para el Banco (cliente). 
</div>
<div align="justify">
- Extraer información de archivos PDF tanto digital como documentos escaneados (data no estructurada), que faciliten la construcción de un *DataSet* para un posterior procesamiento de la data, comunmente conocido como ETL (Extract, Transform, Load), la extracción de la información se hará mediante OCR (Optical Character Recognition) o extracción directa para documentos digitales, se desarrollará un proceso robusto de transformación, limpiando y aplicando reglas que permitan la construcción de un Master Dataset, el cual tendrá como propósito proporcionar métricas y análisis de los informes de peritaje mediante la visualización de un tablero; para toma de decisiones. 
</div>
